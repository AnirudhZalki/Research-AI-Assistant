````python
from mcp_server import MCPServer
from ops import Tracer, FaultInjector, CircuitBreaker, OpsAgent, instrument

connector = MCPServer().connector
tracer = Tracer()
faults = FaultInjector()
instrument(connector, tracer, faults, retries=1, backoff_s=0.001,
           breaker=CircuitBreaker(threshold=2, cooldown_s=60))
faults.arm("search_online_papers", failures=10)
print("Fallback papers:", len(connector.search_online_papers("edge", agent_id="Lab10")))
print("Breaker state:", connector.breaker.state)
print("Diagnosis:", OpsAgent().diagnose(tracer, connector))
OpsAgent().remediate(connector)
print("Remediated state:", connector.breaker.state)
````
